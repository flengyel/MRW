# Quadratic discriminant by traces

See [the working note](README.md) for the definitions and comparison.

```sage
from sage.version import version as sage_version
print(sage_version)
R.<t> = PolynomialRing(QQ)
K.<b> = NumberField(t^2 - t - 1)
basis = [K(1), b]
T = matrix(QQ, [[(u*v).trace() for v in basis] for u in basis])
print(T)
print('Discriminant:', T.det())
```
